```python
import random

def linear_search_count(data, target):
    comparisons = 0
    for idx, val in enumerate(data):
        comparisons = comparisons + 1
        if val == target:
            return idx, comparisons
    return -1, comparisons

def rotated_binary_search(data, target):
    low = 0
    high = len(data) - 1
    comparisons = 0
    
    while low <= high:
        mid = (low + high) // 2
        comparisons = comparisons + 1
        
        if data[mid] == target:
            return mid, comparisons
        
        if data[low] <= data[mid]:
            if data[low] <= target and target < data[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            if data[mid] < target and target <= data[high]:
                low = mid + 1
            else:
                high = mid - 1
    
    return -1, comparisons

random.seed(ID_SUM + SEED_DIGIT)

n_records = 20 + SEED_DIGIT
original_records = sorted(random.sample(range(100, 1000), n_records))

rotation = SEED_DIGIT % n_records
rotated_records = original_records[rotation:] + original_records[:rotation]

target = (ID_SUM + ARTIST_SUM) % 900 + 100

lin_index, lin_comp = linear_search_count(rotated_records, target)
bin_index, bin_comp = rotated_binary_search(rotated_records, target)

print("=" * 55)
print("SEARCH RESULTS — ROTATED RECORD LIST")
print("=" * 55)
print("Target Incident:", target)
print("Dataset Size   :", len(rotated_records))
print("Rotation Point :", rotation)
print("-" * 55)

if lin_index != -1:
    print("Linear Search  → Found at index", lin_index, "|", lin_comp, "comparisons")
else:
    print("Linear Search  → NOT Found |", lin_comp, "comparisons")

if bin_index != -1:
    print("Binary Search  → Found at index", bin_index, "|", bin_comp, "comparisons")
else:
    print("Binary Search  → NOT Found |", bin_comp, "comparisons")

print("-" * 55)
print("Comparison Reduction:", lin_comp - bin_comp, "fewer checks with Binary Search")
print("Speed Factor:", round(lin_comp / bin_comp, 2), "x")
print("=" * 55)
```

```python
import random

random.seed(ID_SUM + SEED_DIGIT)

size_A = 10 + SEED_DIGIT
start_A = ID_SUM % 500
dataset_A = sorted(random.sample(range(start_A, start_A + 500), size_A))

size_B = 10 + ARTIST_LENGTH
start_B = (NAME_LENGTH * 100) % 500
dataset_B = sorted(random.sample(range(start_B, start_B + 500), size_B))

def merge_sorted_lists(list1, list2):
    i = j = comparisons = 0
    merged = []
    while i < len(list1) and j < len(list2):
        comparisons += 1
        if list1[i] < list2[j]:
            merged.append(list1[i])
            i += 1
        else:
            merged.append(list2[j])
            j += 1
    merged.extend(list1[i:])
    merged.extend(list2[j:])
    return merged, comparisons

consolidated, comp_count = merge_sorted_lists(dataset_A, dataset_B)
is_ordered = all(consolidated[k] <= consolidated[k+1] for k in range(len(consolidated)-1))

print("DEPARTMENT A DATASET:")
print(dataset_A)
print()
print("DEPARTMENT B DATASET:")
print(dataset_B)
print()
print("CONSOLIDATED DATASET:")
print(consolidated)
print()
print("COMPARISONS PERFORMED:", comp_count)
print("ORDER VERIFICATION    :", "PASS ✓" if is_ordered else "FAIL ✗")
print()
print("COMPLEXITY ANALYSIS")
print("Time Complexity  : O(n + m) — linear merge")
print("Space Complexity : O(n + m) — new merged list")
print("-" * 60)
```

```python
def bubble_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps

def selection_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps

def insertion_sort(data):
    a, comparisons, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts

data_size = 12 + SEED_DIGIT
base = list(range(100, 100 + data_size))

already_sorted = base.copy()
reverse_order = base[::-1]
unsorted = base.copy()
random.shuffle(unsorted)

def run_all_sorters(dataset, label):
    print(f"--- {label} ---")
    print("Dataset:", dataset)
    b_res, b_comp, b_sw = bubble_sort(dataset)
    s_res, s_comp, s_sw = selection_sort(dataset)
    i_res, i_comp, i_sh = insertion_sort(dataset)
    print(f"Bubble     → Comparisons: {b_comp}, Swaps: {b_sw}")
    print(f"Selection  → Comparisons: {s_comp}, Swaps: {s_sw}")
    print(f"Insertion  → Comparisons: {i_comp}, Shifts: {i_sh}")
    print()
    return (b_comp, s_comp, i_comp)

print("E-COMMERCE ORDER PROCESSING COMPARISON")
print("-" * 60)
run_all_sorters(already_sorted, "SCENARIO 1: Already Sorted")
run_all_sorters(reverse_order,    "SCENARIO 2: Reverse Order")
run_all_sorters(unsorted,         "SCENARIO 3: Random/Unsorted")

print("SUMMARY OBSERVATIONS")
print("- Already sorted: Insertion & Bubble fastest (early exit)")
print("- Reverse order: All slowest — worst-case for all O(n²) sorts")
print("- Random order: Performance falls between best & worst cases")
print("- Selection Sort: Consistent comparisons, fewer swaps overall")
print("-" * 60)
```

```python
import random
import time

def bubble_sort(data):
    a = data.copy()
    comparisons = 0
    swaps = 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons = comparisons + 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps = swaps + 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps

def selection_sort(data):
    a = data.copy()
    comparisons = 0
    swaps = 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons = comparisons + 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps = swaps + 1
    return a, comparisons, swaps

def insertion_sort(data):
    a = data.copy()
    comparisons = 0
    shifts = 0
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0:
            comparisons = comparisons + 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts = shifts + 1
            j = j - 1
        a[j + 1] = key
    return a, comparisons, shifts

def measure_sort(sort_func, data):
    start = time.perf_counter()
    result = sort_func(data)
    elapsed = time.perf_counter() - start
    return result + (elapsed,)
```

```python
random.seed(ID_SUM + SEED_DIGIT)
data_size = 10 + SEED_DIGIT
base_orders = [random.randint(100, 999) for _ in range(data_size)]

sorted_input = sorted(base_orders)
reverse_input = sorted(base_orders, reverse=True)
random_input = base_orders.copy()
random.shuffle(random_input)

datasets = [
    ("Already Sorted", sorted_input),
    ("Reverse Order", reverse_input),
    ("Random/Unsorted", random_input)
]

print("E-COMMERCE ORDER PROCESSING COMPARISON")
print("Dataset Size:", data_size)
print("-" * 70)

for label, data in datasets:
    print("\nSCENARIO:", label)
    print("Input Data:", data)
    
    b_res, b_comp, b_mov, b_time = measure_sort(bubble_sort, data)
    s_res, s_comp, s_mov, s_time = measure_sort(selection_sort, data)
    i_res, i_comp, i_mov, i_time = measure_sort(insertion_sort, data)
    
    print("Bubble Sort    → Comparisons:", b_comp, "| Movements:", b_mov, "| Time:", round(b_time, 8), "s")
    print("Selection Sort → Comparisons:", s_comp, "| Movements:", s_mov, "| Time:", round(s_time, 8), "s")
    print("Insertion Sort → Comparisons:", i_comp, "| Movements:", i_mov, "| Time:", round(i_time, 8), "s")
    
    all_sorted = (b_res == s_res == i_res == sorted(data))
    print("Correctly Sorted:", "✓ PASS" if all_sorted else "✗ FAIL")

print("\n" + "-" * 70)
print("BIG O & CASE ANALYSIS")
print("-" * 70)
print("Best-Case (already sorted):")
print("  Bubble Sort   → O(n) — early exit when no swaps")
print("  Insertion Sort → O(n) — no shifting needed")
print("  Selection Sort → O(n²) — always scans fully")
print()
print("Worst-Case (reverse order):")
print("  All O(n²) — maximum comparisons & moves required")
print()
print("Average-Case (random order):")
print("  All O(n²) — performance falls between best & worst")
print()
print("Observation: Insertion Sort is fastest in practice;")
print("Selection Sort has fewest moves; Bubble Sort is slowest.")
```
